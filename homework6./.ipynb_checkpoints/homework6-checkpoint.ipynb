```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, n):
        if n % i == 0:     
            return False
    return True
```

```python
def prime_rows(arr):
    prime = []
    for row in arr:
        for value in row:
            if is_prime(value):
                prime.append(row)    # error, the code kept going so it printed the row for each prime in it
                break # this is what google said to do as I was stumped
    return np.array(prime)
```

```python
def checkers():
    board = np.zeros((8, 8), dtype=int)
    for i in range(board.shape[0]):
        for j in range(board.shape[1]):
            if (i + j) % 2 == 0:
                board[i, j] = 1
    return board

print(checkers())
```

```python
def reverse_checkers():
    board = np.zeros((8, 8), dtype=int)
    for i in range(board.shape[0]):
        for j in range(board.shape[1]):
            if (i + j) % 2 == 1:
                board[i, j] = 1
    return board

print(reverse_checkers())
```

```python
def kaboom(words, n):
    spacer = " " * n
    result = []
    for word in words:
        result.append(spacer.join(word))
    return np.array(result)


universe = np.array(['galaxy', 'clusters'])
print(kaboom(universe, 1))
print(kaboom(universe, 2))
print(kaboom(universe, 3))
print(kaboom(universe, 80))
```

```python
def onlysecondbest(stars):
    sorted_stars = np.sort(stars, axis=1)
    return sorted_stars[:, -2]


stars = np.array([[1123, 1456, 1789, 1324, 1876],
                  [1567, 1987, 1678, 1405, 1589],
                  [1345, 1654, 1523, 1109, 1923],
                  [1298, 1890, 1367, 1784, 1432],
                  [1823, 1756, 1489, 1672, 1550]])

print(onlysecondbest(stars)) # error: I accidentally typed "only second best(stars)"
```

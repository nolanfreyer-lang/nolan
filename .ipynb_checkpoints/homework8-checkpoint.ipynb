```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x, A, w):
    return A * np.sin(w * x)
```

```python
x = np.linspace(0, 2 * np.pi, 1000)

amp = [0.5, 1, 1.5, 2, 2.5]
freq = [1, 2, 3, 4, 5]
```

```python
fig, axes = plt.subplots(1, 3, figsize=(25, 5))

for A, w, in zip(amp, freq):
    axes[0].plot(x, make_sine_wave(x, A, 1), label=f'A = {A}, w = 1')
    axes[1].plot(x, make_sine_wave(x, 1, w), label=f'A = 1, w = {w}')
    axes[2].plot(x, make_sine_wave(x, A, w), label=f'A = {A}, w = {w}')

titles = ['varying amp', 'varying freq', 'varying amp and freq']
for ax, title in zip(axes, titles):
    ax.set_title(title, fontsize=18)
    ax.set_xlabel('x', fontsize=14)
    ax.set_ylabel('y', fontsize=14)
    ax.legend(fontsize=12)

plt.show()
```

```python
med_lat = df['latitude'].median()
med_lon = df['longitude'].median()
print(med_lat, med_lon)
```

```python
df['latitude distance'] = (df['latitude'] - med_lat).abs()
df['longitude distance'] = (df['longitude'] - med_lon).abs()
```

```python
df = df.drop(columns=['lat_dist', 'lon_dist'])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
penguins = sns.load_dataset('penguins')
```

```python
penguins = penguins.dropna(axis=1)
```

```python
# species of penguin, island they live on, bill length, bill depth, flipper length, weight, sex
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

sns.scatterplot(data=penguins, x='bill_length_mm', y='bill_depth_mm',
                hue='species', ax=axes[0, 0])
axes[0, 0].set_title('Bill Length vs. Bill Depth')
axes[0, 0].set_xlabel('Bill Length (mm)')
axes[0, 0].set_ylabel('Bill Depth (mm)')

sns.scatterplot(data=penguins, x='flipper_length_mm', y='body_mass_g',
                color='green', ax=axes[0, 1])
axes[0, 1].set_title('Flipper Length vs. Body Mass')
axes[0, 1].set_xlabel('Flipper Length (mm)')
axes[0, 1].set_ylabel('Body Mass (g)')

sns.histplot(data=penguins, x='body_mass_g', color='green', ax=axes[1, 0])
axes[1, 0].set_title('Distribution of Body Mass')
axes[1, 0].set_xlabel('Body Mass (g)')
axes[1, 0].set_ylabel('Count (number of penguins)')

sns.boxplot(data=penguins, x='island', y='flipper_length_mm',
            color='green', ax=axes[1, 1])
axes[1, 1].set_title('Flipper Length by Island')
axes[1, 1].set_xlabel('Island')
axes[1, 1].set_ylabel('Flipper Length (mm)')
# greeeeeeeeeeeeeeeeen
plt.show()
```
